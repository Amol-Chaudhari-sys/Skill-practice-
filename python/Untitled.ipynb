```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate 400 evenly spaced x-values between -10 and 10
x = np.linspace(-10, 10, 100)

# 2. Define your equations
y1 = x  
y2 = x+1/(x*x)

# 3. Create the plot
plt.figure(figsize=(8, 6))
plt.plot(x, y1, label='y = 2x + 3', color='blue')
plt.plot(x, y2, label='y = -0.5x + 1', color='red')

# 4. Customize the chart layout
plt.axhline(0, color='black', linewidth=0.5)  # X-axis line
plt.axvline(0, color='black', linewidth=0.5)  # Y-axis line
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.title('Plot of Linear Equations')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()  # Displays the equation labels

# 5. Show the window
plt.show()
```
