# mean absulate divusion 

```python
sec_a = np.array([75,65,73,68,72,67])
sec_b =np.array([90,47,43,96,93,51])
no = np.array([1,2,3,4,5,6])
```

```python
mn_a = np.mean(sec_a)
mn_b=np.mean(sec_b)
print (mn_a,mn_b)
```

```python
plt.figure(figsize=(10,3))

plt.scatter(sec_a,no,color="red" ,label="sec a")
plt.scatter(sec_b,no,color="green",label ="sec b")
plt.plot ([mn_a for i in range (0,6)],no,c='black',label="mean")
plt.legend()
plt.show()
```

```python


mad_a = np.sum (np.abs ( sec_a - mn_a )) / len ( sec_a) 

# mean absulate division 

mad_b=np.sum ( np.abs ( sec_b - mn_b ) ) / len (sec_b)
```

```python
mad_a,mad_b
```
