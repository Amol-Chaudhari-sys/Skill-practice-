```python
def printnum (i, n ):
    if i>n:
        return 
    print (i , end=" ")
    printnum(i+1,n)
```

```python
def fact (n ):
    if n ==0 :
        return 1
    return n*fact(n-1)
```

```python
print(fact(5))
```

```python
def gcd (a,b):
    if b ==0:
        return a
    return gcd(b,a%b)
```

```python
gcd(15,50)
```

```python
def lcm (a,b ):
    return a*b/gcd(a,b)
```

```python
lcm (15,50)
```
