```python
from heapq import heappush ,heappop 
def dijkastras (n , edges , s ):
    adjlist = []
    for i in range (n ):
        adjlist.append ([])
    for edge in edges:
        adjlist[edge[0]].append([edge[1],edge[2]])
        adjlist[edge[1]].append([edge[0],edge[2]])

    
    dist =[float('inf')]*n 
    heap =[]
    dist[s]=0
    heappush(heap, [dist[s],s])
    while len (heap) >0 :
        d, u= heappop(heap)
        for v, w in adjlist [u]:
            if w+dist[u] < dist [v]:
                dist [v]= w+dist [u]
                heappush (heap , [dist[v], v])
    return dist
dist = dijkastras (6, edges, 0 )
```
