# sorting 

```python
nums = [5,2,3,12,2,1,6,9,7]

```

```python
def bubble (num  ):
    n= len(num)
    for i in range (n ):
        isswap =False
        for j in range (n-i-1):
            if num[j]>num[j+1]:
                num[j],num[j+1]=num[j+1],num[j]
                isswap=True 
        if not isswap:
            break;
    return num
bubble(nums)
```

```python
def insertion (num):
    n= len (num)
    for i in range(1,n):
        key  = num[i]
        j=i-1
        while j>=0 and num[j]>key :
            num[j+1]=num[j ]
            j-=1
        num[j+1]=key 
    return num
insertion (nums)
```

```python
def selection(nums):
    n=len (nums)
    for i in range (n ):
        mn = nums[i]
        ind=i
        for j in range (i+1,n ):
            if nums[j]<mn :
                mn=nums[j]
                ind=j
        nums[i],nums[ind]=nums[ind],nums[i]
    return nums
selection(nums)
```

```python
# merge sort 
# tc =nlog n 
def merge (nums, l ,mid ,r ):
    l1=[]
    l2=[]
    for i in range (l,mid+1):
        l1.append(nums[i])
 
    for i in range (mid+1, r+1):
        l2.append (nums[i])
      
    i,j,k=0,0,l
    while k<=r:
        if i==len(l1):
            nums[k]=l2[j]
            j+=1
            k+=1
        elif(j==len(l2)):
            nums[k]=l1[i]
            i+=1
            k+=1
        elif l1[i]<l2[j]:
            nums[k]=l1[i]
            i+=1
            k+=1
        else:
            nums[k]=l2[j]
            j+=1
            k+=1
    
def mergesort (nums, l ,r ):
    if l>=r:
        return 
    
    mid = (l+r)//2 
    
    mergesort (nums, l , mid)
    mergesort (nums, mid+1, r )
    merge(nums, l , mid ,r)
def mgsort(nums):
    mergesort(nums,0 , len (nums)-1)
    return nums
```

```python
mgsort(nums)
```

```python
# quick sort 
def partation (nums, l , r ):
    key= nums[r]
    start =l
    for i in range (l,r+1):
        if nums[i]<=key:
            nums[start], nums[i]=nums[i],nums[start]
            start+=1
    return start-1
    
def quicksort (nums, l , r ):
    if l>=r:
        return 
    p = partation (nums,l,r)
    
    quicksort(nums, l , p-1)
    quicksort(nums, p+1, r )
def qsort(nums):
    quicksort(nums, 0 , len (nums)-1)
    return nums
```

```python
qsort(nums)
```

```python
# counting sort 
def countingsort(nums):
    print ("bafore ", nums)
    print ("after ", end="")
    n=len (nums)
    mx=max(nums)
    freq=[0]*(mx+1)
    for i in nums:
        freq[i]+=1
    nums=[]
    for i in range (0,mx+1):
        while freq[i]>0:
            nums.append(i)
            freq[i]-=1
    return nums
countingsort(nums)
```

```python
# lower bound 
def lowerbound(nums, target):
    l=0
    r=len(nums)-1
    ans=len (nums)
    while l<=r:
        mid = (l+r)//2
        if nums[mid]>=target :
            ans=mid
            r=mid-1
        else:
            l=mid+1
    return ans
lowerbound(nums, 16)
```

```python
def upperbound(nums,target):
    l=0
    r=len (nums)-1
    ans=len (nums)
    while l<=r:
        mid = (l+r)//2
        if nums[mid ]>target :
            ans=mid
            r=mid-1
        else :
            l=mid+1
    return ans
upperbound(nums, 16)
```
