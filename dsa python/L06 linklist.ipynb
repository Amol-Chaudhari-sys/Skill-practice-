# circular link list 


```python
class Node :
    def __init__(self ,data):
        self.data= data
        self.next=None
```

```python
a=Node(5)
b=Node(3)
c=Node(7)
d=Node(8)

a.next=b
b.next=c
c.next=d
d.next = a
head= a
```

```python
current= head 
while True :
    print (current.data ,end = " ")
    current = current.next
    if current == head :
        break 
```
