```python
class stack:
    def __init__(self):
        self.st=[]
    def push (self,x):
        self.st.append (x)
    def pop (self ):
        if (len(self.st)==0):
            return -1
        self.st.pop()
        return self.st[-1]
    def top(self ):
        if len (self.st)==0 :
            return -1
        return self.st[-1]
    def size (self ):
        return len (self.st)
```

```python
st= stack()
```

```python
st.push (7)

```

```python
st.pop()
```

```python
st.top()
```
