```python
# post order
def postorder(root):
    if root is None :
        return 
    postorder(root.rigth)
    postorder(root.left)
    ans.append(root.data)
```

```python
def bsearch(root,target ):
    if root ==None :
        return None
    curr= root
    while curr!= None :
        if curr.val == target :
            return curr
        elif (curr.val > target ):
            curr = curr.left
        else :
            curr= curr.right 
    return None 
```
