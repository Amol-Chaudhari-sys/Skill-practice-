```python
import matplotlib.pyplot  as plt
import numpy as np 
```

```python
import random
```

```python
x= []
y = []
while len (x)<100:
    x.append(random.randint(1,100))
    y.append(random.randint(2, 300))
print (x)
print (y)
```

```python
x= np.array(x)
y = np.array(y)
plt.plot(x, marker= 'o', linestyle="dotted" , color= 'orange')
plt.plot (y, marker= '*', color= 'y')
plt.xlabel ("x - axis ")
plt.ylabel("y - axis ")
plt.grid(linewidth = 1)

```

```python
plt.bar(x,y)
```
